```python
import pandas as pd
```

```python
df = pd.read_csv("../data/enriched_dataset.csv")

df.head()
```

```python
print("Dataset shape:", df.shape)

print("\nColumns:")
print(df.columns)

print("\nMissing transcripts:", df["transcript"].isna().sum())
```

```python
import re

def clean_text(text):
    if pd.isna(text):
        return text
    
    # remove newline and tab characters
    text = re.sub(r'\n', ' ', text)
    text = re.sub(r'\t', ' ', text)
    
    # remove special characters
    text = re.sub(r'#|@|\*', '', text)
    
    # remove multiple spaces
    text = re.sub(r'\s+', ' ', text)

    return text.strip()

# apply cleaning
df["title"] = df["title"].apply(clean_text)
df["transcript"] = df["transcript"].apply(clean_text)

df.head()
```

```python
df = df.dropna(subset=["transcript"])

print("Remaining rows:", len(df))
```

For semantic search we only need:

video_id
title
published_date
transcript

```python
df = df[["video_id", "title", "published_date", "transcript"]]

df.head()
```

```python
df.to_csv("../data/cleaned_transcripts.csv", index=False)

print("Cleaned transcripts dataset saved successfully!")
```

```python
from sentence_transformers import SentenceTransformer
```

```python
model = SentenceTransformer("all-MiniLM-L6-v2")
```

```python
sentences = [
    "What is Python?",
    "Python is a programming language"
]

embeddings = model.encode(sentences)

print("Embedding shape:", embeddings.shape)
```

```python
from sentence_transformers import SentenceTransformer

models = [
    "all-MiniLM-L6-v2",
    "multi-qa-MiniLM-L6-cos-v1",
    "all-mpnet-base-v2"
]

sentences = [
    "What is Python programming?",
    "Explain machine learning"
]

for m in models:
    print("\nTesting model:", m)
    
    model = SentenceTransformer(m)
    
    embeddings = model.encode(sentences)
    
    print("Embedding shape:", embeddings.shape)
```
