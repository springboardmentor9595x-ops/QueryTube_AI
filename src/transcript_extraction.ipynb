```python
import pandas as pd
from youtube_transcript_api import YouTubeTranscriptApi
from youtube_transcript_api._errors import TranscriptsDisabled, NoTranscriptFound
```

```python
df = pd.read_csv("../data/cleaned_metadata.csv")

print("Dataset Loaded Successfully")
print("Columns:", df.columns)

if "video_id" not in df.columns:
    raise ValueError("video_id column not found in dataset")
```

```python
from youtube_transcript_api import YouTubeTranscriptApi
from youtube_transcript_api._errors import TranscriptsDisabled, NoTranscriptFound
import time

def get_transcript(video_id):
    try:
        # Try normal transcript
        transcript = YouTubeTranscriptApi.get_transcript(video_id)
        text = " ".join([seg["text"] for seg in transcript])
        return text

    except NoTranscriptFound:
        try:
            # Try auto generated captions
            transcript_list = YouTubeTranscriptApi.list_transcripts(video_id)
            transcript = transcript_list.find_generated_transcript(['en'])
            data = transcript.fetch()
            text = " ".join([seg["text"] for seg in data])
            return text

        except:
            return None

    except TranscriptsDisabled:
        return None

    except Exception:
        return None
```

```python
transcripts = []
failures = []

for video_id in df["video_id"]:
    
    text = get_transcript(video_id)

    if text:
        transcripts.append(text)
    else:
        transcripts.append(None)
        failures.append(video_id)

    time.sleep(0.5)   # prevents YouTube blocking
```

```python
df["transcript"] = transcripts
```

```python
df.to_csv("../data/video_with_transcripts.csv", index=False)

import pandas as pd
fail_df = pd.DataFrame(failures, columns=["video_id"])
fail_df.to_csv("../data/transcript_failures.csv", index=False)
```

```python
print("Total videos:", len(df))
print("Transcripts found:", df["transcript"].notna().sum())
print("Failures:", df["transcript"].isna().sum())
```
