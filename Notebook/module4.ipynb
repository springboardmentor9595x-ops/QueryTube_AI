```python
# Step 1: Load the Transcript Dataset

import pandas as pd

# Load the dataset
df = pd.read_csv("../data/enriched_dataset.csv") 

# Display first 5 rows
print("First 5 rows of the dataset:")
print(df.head())

# Print dataset shape
print("\nDataset Shape:", df.shape)

# Print column names
print("\nColumn Names:")
print(df.columns)

# Display dataset information
print("\nDataset Info:")
print(df.info())
```

```python
import re

def clean_text(text):
    if isinstance(text, str):
        text = text.replace('\n', ' ')
        text = text.replace('\t', ' ')
        text = re.sub(r'[#@*]', '', text)
        text = re.sub(r'[^\w\s]', '', text)
        text = re.sub(r'\s+', ' ', text)
        text = text.strip()
    return text

df['title'] = df['title'].apply(clean_text)
df['transcript'] = df['transcript'].apply(clean_text)

print(df[['title', 'transcript']].head())
```

```python
# Step 3: Handle Missing or Null Transcripts

# Remove rows where transcript is NaN
df = df.dropna(subset=['transcript'])

# Remove rows where transcript is empty
df = df[df['transcript'].str.strip() != ""]

# Reset index after removal
df = df.reset_index(drop=True)

# Check dataset shape after cleaning
print("Dataset Shape after removing null transcripts:", df.shape)
```

```python
# Rename published_date to datetime first
df = df.rename(columns={'publish_date': 'datetime'})

# Remove unnecessary columns
df = df.drop(columns=['eeror', 'year', 'month', 'title_length'], errors='ignore')

# Keep only required columns
df = df[['video_id', 'title', 'datetime', 'transcript']]

# Convert datetime
df['datetime'] = pd.to_datetime(df['datetime'], errors='coerce')
```

```python
# Step 4: Normalize the Dataset Format

# Check current columns
print("Current columns:", df.columns)

# Rename published_date to datetime if present
if 'published_date' in df.columns:
    df = df.rename(columns={'published_date': 'datetime'})
elif 'publish_date' in df.columns:
    df = df.rename(columns={'publish_date': 'datetime'})

# Remove unnecessary columns if they exist
df = df.drop(columns=['eeror', 'year', 'month', 'title_length'], errors='ignore')

# Keep only required columns
df = df[['video_id', 'title', 'datetime', 'transcript']]

# Convert datetime column to proper datetime format
df['datetime'] = pd.to_datetime(df['datetime'], errors='coerce')

# Save cleaned dataset
df.to_csv("cleaned_transcripts.csv", index=False)

# Preview dataset
print(df.head())
print("Final Dataset Shape:", df.shape)
```

```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# Load transcripts dataset
videos = pd.read_csv("../Notebook/cleaned_transcripts.csv")

# Read queries from txt file
with open("../data/queries.txt", "r", encoding="utf-8") as f:
    query_list = [line.strip() for line in f if line.strip()]

# Convert queries to dataframe
queries = pd.DataFrame(query_list, columns=["query"])

# Combine title + transcript for better matching
videos["content"] = videos["title"] + " " + videos["transcript"]

# Create TF-IDF vectors
vectorizer = TfidfVectorizer(stop_words="english")

video_vectors = vectorizer.fit_transform(videos["content"])
query_vectors = vectorizer.transform(queries["query"])

# Compute similarity
similarity = cosine_similarity(query_vectors, video_vectors)

# Find best matching video for each query
best_match_index = similarity.argmax(axis=1)

# Map queries to video_id
queries["relevant_video_id"] = videos.iloc[best_match_index]["video_id"].values

# Save mapping
queries.to_csv("query_video_mapping.csv", index=False)

print(queries.head())
```
